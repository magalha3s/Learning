## Importing Pytorch

```python
import torch 
torch.__version__
```

## Tensors

```python
# Scalar
scalar = torch.tensor(7)

# Vector
vector = torch.tensor([7,7])

# Matrix
matrix = torch.tensor([[7,8],[9,10]])

# Tensor
tensor = torch.tensor([[[1,2,3],[3,6,9],[2,4,5]]])
```

## Random Tensors

```python
# Tensor with random entries
random_tensor = torch.rand(size=(224,224,3))

# Tensor with only zeros
zeros = torch.zeros(size=(3,4))

# Tensor with only ones
ones = torch.ones(size=(3,4))

# Use torch.arange() torch.range() is deprecated
zero_to_ten = torch.arange(0,10,1)

# Copy the shape of a tensor to another tensor with zeros_like() or ones_like()
ten_zeros = torch.zeros_like(input=zero_to_ten)
```

## Manipulating Tensors

```python
# Basic Operations
tensor = torch.tensor([1,2,3])
tensor + 10
tensor * 10
tensor - 10

# Shapes need to be in the right way  
tensor_A = torch.tensor([[1, 2], [3, 4], [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10], [8, 11], [9, 12]], dtype=torch.float32)

# Finding the min, max, mean, sum, etc
x = torch.arange(0,100,10)
x.min(), x.max(), x.type(torch.float32).mean(), x.sum()

torch.Tensor.view(x)
```
